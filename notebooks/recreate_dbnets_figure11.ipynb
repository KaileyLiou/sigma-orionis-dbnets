# Reproducing Figure 11 from the DBNets 2.0 Paper
This notebook recreates Figure 11 from the DBNets 2.0 publication using the published planet masses and orbital distances.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/kaileyliou/Documents/sigma-orionis-dbnets/data/Table E.1 Data - Sheet1.csv")

distance = df["a_au"]
mass = df["Mp_MJ"]

lower_error = df["Mp_lower"]
upper_error = df["Mp_upper"]

plt.figure(figsize=(8,6))

plt.errorbar(distance, mass, yerr=[lower_error, upper_error], fmt='o', markersize=5, capsize=3)

plt.xscale("log")
plt.yscale("log")

plt.xlim(1e-2, 1e3)
plt.ylim(1e-3, 1e2)

plt.xlabel("Semi-major orbital axis [au]")
plt.ylabel("Planet mass [$M_J$]")

plt.tight_layout()
plt.show()
```
