# Reproducing Figure 11 from the DBNets 2.0 Paper
This notebook recreates Figure 11 from the DBNets 2.0 publication using the published planet masses and orbital distances.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/kaileyliou/Documents/sigma-orionis-dbnets/data/DBNets 2.0 Data - Sheet1.csv")

distance = df["a_au"]
value = df["Mp_MJ"]

lower_error = df["Mp_lower"]
upper_error = df["Mp_upper"]

solar_distance = [0.387, 0.723, 1.000, 1.524, 5.203, 9.537, 19.191, 30.069]
solar_mass = [0.00017, 0.00245, 0.00315, 0.00034, 1.0, 0.299, 0.0457, 0.0539]

plt.figure(figsize=(9,5))

plt.errorbar(distance, value, yerr=[lower_error, upper_error], fmt='o', markersize=4, capsize=3, label="DBNets2.0", color='red')

plt.scatter(solar_distance, solar_mass, marker="*", s=120, label="Solar system planets")

sigma_orionis = pd.read_csv("/Users/kaileyliou/Documents/sigma-orionis-dbnets/data/Sigma Orionis DBNets Values - Sheet1.csv")
plt.errorbar(sigma_orionis["Rp (au)"], sigma_orionis["Mp"], yerr=[sigma_orionis["Mp Error -"], sigma_orionis["Mp Error +"]], fmt='o', markersize=6, capsize=3, label="Sigma Orionis", color='green')

plt.legend()

plt.xscale("log")
plt.yscale("log")

plt.xlim(1e-2, 1e3)
plt.ylim(1e-3, 1e2)

plt.xlabel("Semi-major orbital axis [au]")
plt.ylabel("Planet mass [$M_J$]")

plt.title("DBNets 2.0: Planet Mass vs. Semi-major Orbital Axis")

plt.tight_layout()
plt.show()
```
